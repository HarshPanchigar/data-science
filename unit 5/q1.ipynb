```python
import pandas as pd
import matplotlib.pyplot as plt

# Create the monthly sales dataset
data = {
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
              "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"],
    "Sales": [125000, 138000, 142000, 135000, 155000, 162000,
              158000, 171000, 168000, 185000, 215000, 245000],
    "Orders": [820, 910, 950, 890, 1020, 1080,
               1050, 1140, 1110, 1240, 1480, 1690]
}

df = pd.DataFrame(data)

# a) Create a line plot to show the monthly sales trend.
# Use markers and suitable axis labels.
plt.figure(figsize=(10, 5))
plt.plot(df["Month"], df["Sales"], marker="o", label="Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales Amount")
plt.title("Monthly Sales Trend")
plt.legend()

# d) Rotate X-axis month labels by 45 degrees.
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


# b) Create a bar chart showing monthly sales.
plt.figure(figsize=(10, 5))
plt.bar(df["Month"], df["Sales"], label="Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales Amount")
plt.title("Monthly Sales")
plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


# c) Create a line plot showing the monthly number of orders.
plt.figure(figsize=(10, 5))
plt.plot(df["Month"], df["Orders"], marker="o", label="Monthly Orders")
plt.xlabel("Month")
plt.ylabel("Number of Orders")
plt.title("Monthly Orders Trend")

# d) Rotate X-axis month labels by 45 degrees and add a suitable legend.
plt.xticks(rotation=45)
plt.legend()
plt.tight_layout()
plt.show()
```

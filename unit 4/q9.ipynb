```python
import pandas as pd
import numpy as np

# Create sample water consumption data with missing readings
data = {
    "Date": [
        "2026-10-01", "2026-10-02", "2026-10-03",
        "2026-10-04", "2026-10-05", "2026-10-06",
        "2026-10-07", "2026-10-08"
    ],
    "Consumption": [100, 120, np.nan, np.nan, 180, 200, np.nan, 240]
}

df = pd.DataFrame(data)

# Save the sample data as a CSV file
df.to_csv("water_consumption.csv", index=False)

# a. Import the dataset.
df = pd.read_csv("water_consumption.csv", parse_dates=["Date"])

print("Original Dataset:")
print(df)


# b. Identify missing consumption values.
print("\nMissing Consumption Values:")
print(df[df["Consumption"].isnull()])


# d. Store the original values for dates with missing readings.
affected_dates = df.loc[
    df["Consumption"].isnull(), ["Date", "Consumption"]
].copy()


# c. Use interpolate() to estimate the missing readings.
df["Consumption"] = df["Consumption"].interpolate()

print("\nUpdated Dataset After Interpolation:")
print(df)


# d. Display the original and updated values for the affected dates.
affected_dates["Updated_Consumption"] = df.loc[
    affected_dates.index, "Consumption"
]

print("\nOriginal and Updated Values for Affected Dates:")
print(affected_dates)
```

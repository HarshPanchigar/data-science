```python
import pandas as pd
import numpy as np

# Create sample bank customer data with missing values
data = {
    "Customer_ID": [101, 102, 103, 104, 105, 106, 107, 108],
    "Age": [25, 30, 45, np.nan, 35, 40, 28, 50],
    "Income": [30000, np.nan, 50000, 45000, np.nan, 60000, 35000, 55000],
    "Balance": [5000, 8000, np.nan, 12000, 7000, 15000, 9000, 11000],
    "Credit_Score": [650, 700, np.nan, 720, 680, np.nan, 710, 690]
}

df = pd.DataFrame(data)

# Save the sample data to an Excel file
# df.to_excel("bank_customers.xlsx", index=False)

# a. Import the Excel file.
# df = pd.read_excel("bank_customers.xlsx")

print("Original Dataset:")
print(df)

# b. Find the number of missing values in every column.
print("\nMissing Values in Each Column:")
print(df.isnull().sum())

# c. Identify the column having the maximum number of missing values.
missing_counts = df.isnull().sum()

print("\nColumn with Maximum Missing Values:")
print(missing_counts.idxmax())

# d. Replace missing Income values with the median income.
median_income = df["Income"].median()
df["Income"] = df["Income"].fillna(median_income)

print("\nMedian Income:", median_income)

# e. Replace missing Credit_Score values with the mean credit score.
mean_credit_score = df["Credit_Score"].mean()
df["Credit_Score"] = df["Credit_Score"].fillna(mean_credit_score)

print("Mean Credit Score:", mean_credit_score)

# f. Verify the changes.
print("\nDataset After Replacing Missing Values:")
print(df)

print("\nMissing Values After Replacement:")
print(df.isnull().sum())
```

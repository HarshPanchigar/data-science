```python

import pandas as pd

# Create the DataFrame
data = {
    "Calories_Burned": [
        450, 520, 480, 510, 490,
        530, 475, 500, 495, 1200
    ]
}

df = pd.DataFrame(data)

# a. Calculate mean and median
print("Mean:", df["Calories_Burned"].mean())
print("Median:", df["Calories_Burned"].median())

# b. Calculate variance and standard deviation
print("Variance:", df["Calories_Burned"].var())
print("Standard Deviation:", df["Calories_Burned"].std())

# c. Identify the more representative measure
print("More representative measure: Median")

# d. Compare mean with and without 1200
df_without_1200 = df[df["Calories_Burned"] != 1200]

print("Mean with 1200:", df["Calories_Burned"].mean())
print("Mean without 1200:", df_without_1200["Calories_Burned"].mean())
```

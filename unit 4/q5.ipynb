```python
import pandas as pd

# 1. Create a retail sales dataset.
data = {
    "Product": ["Laptop", "Phone", "Chair", "Table", "Tablet",
                "Headphones", "Monitor", "Keyboard", "Printer", "Camera"],
    "Category": ["Electronics", "Electronics", "Furniture", "Furniture",
                 "Electronics", "Electronics", "Electronics", "Accessories",
                 "Electronics", "Electronics"],
    "Quantity": [2, 5, 4, 2, 3, 10, 4, 8, 2, 3],
    "Unit_Price": [50000, 15000, 3000, 8000, 20000,
                   2000, 12000, 1000, 10000, 25000],
    "Total_Sales": [100000, 75000, 12000, 16000, 60000,
                    20000, 48000, 8000, 20000, 75000]
}

df = pd.DataFrame(data)

# Save the dataset as an Excel file.
# df.to_excel("retail_sales.xlsx", index=False)

# a. Import the Excel dataset.
# df = pd.read_excel("retail_sales.xlsx")

print("Retail Sales Dataset:")
print(df)

# b. Calculate Q1 and Q3 of Total_Sales.
Q1 = df["Total_Sales"].quantile(0.25)
Q3 = df["Total_Sales"].quantile(0.75)

print("\nQ1:", Q1)
print("Q3:", Q3)

# c. Calculate the Interquartile Range (IQR).
IQR = Q3 - Q1

print("IQR:", IQR)

# d. Calculate cthe lower and upper limits for detecting outliers.
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)

# e. Display transactions whose Total_Sales values are outside these limits.
outliers = df[
    (df["Total_Sales"] < lower_limit) |
    (df["Total_Sales"] > upper_limit)
]

print("\nTransactions with Outliers:")
print(outliers)
```

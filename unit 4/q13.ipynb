```python
import pandas as pd

# Create sample hotel booking data
data = {
    "Booking_Date": [
        "2023-01-15", "2023-03-20", "2023-06-10",
        "2023-09-05", "2024-02-12", "2024-04-18",
        "2024-07-22", "2024-11-30", "2025-01-10",
        "2025-05-15", "2025-08-20", "2025-12-25"
    ],
    "Room_Type": [
        "Standard", "Deluxe", "Suite", "Standard",
        "Deluxe", "Suite", "Standard", "Deluxe",
        "Suite", "Standard", "Deluxe", "Suite"
    ],
    "Guests": [2, 3, 4, 2, 3, 4, 2, 3, 4, 2, 3, 4],
    "Stay_Days": [2, 3, 5, 1, 4, 6, 2, 3, 5, 2, 4, 7],
    "Booking_Amount": [
        5000, 9000, 20000, 4000,
        12000, 25000, 6000, 11000,
        22000, 7000, 15000, 30000
    ]
}

df = pd.DataFrame(data)

# Save the sample data as a CSV file
# df.to_csv("hotel_bookings.csv", index=False)

# a. Convert Booking_Date into datetime format.
df["Booking_Date"] = pd.to_datetime(df["Booking_Date"])

# b. Extract the year and month.
df["Year"] = df["Booking_Date"].dt.year
df["Month"] = df["Booking_Date"].dt.month

print("Hotel Booking Dataset:")
print(df)

# c. Calculate the number of bookings for each year using count().
bookings_per_year = df.groupby("Year")["Booking_Date"].count()

print("\nNumber of Bookings for Each Year:")
print(bookings_per_year)

# d. Calculate the average booking amount for each year.
average_amount = df.groupby("Year")["Booking_Amount"].mean()

print("\nAverage Booking Amount for Each Year:")
print(average_amount)

# e. Calculate the highest booking amount for each year.
highest_amount = df.groupby("Year")["Booking_Amount"].max()

print("\nHighest Booking Amount for Each Year:")
print(highest_amount)

# f. Determine which year generated the highest average booking amount.
highest_average_year = average_amount.idxmax()

print("\nYear with Highest Average Booking Amount:", highest_average_year)
print("Highest Average Amount:", average_amount.max())
```

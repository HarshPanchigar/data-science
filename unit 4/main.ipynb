```python
data = {
    "Order_ID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "Age": [22, 35, 28, 45, 19, 32, 40, 25, 50, 30],
    "Distance_km": [3.5, 5.2, 2.8, 8.0, 1.5, 6.3, 4.5, 7.2, 9.0, 3.0],
    "Order_Value": [250, 450, 180, 600, 150, 520, 350, 700, 800, 300],
    "Delivery_Time": [25, 40, 20, 55, 15, 45, 35, 60, 65, 30],
    "Rating": [5, 4, 5, 3, 5, 4, 4, 2, 3, 5]
}

df = pd.DataFrame(data)
print(df)
print("Mean Age:", df["Age"].mean())
print(df["Age"].mode().tolist())

mean_age = df["Age"].mean()
median_age = df["Age"].median()

print("\nComparison of Age:")
if abs(mean_age - median_age) <= 2:
    print("Mean and median are approximately similar.")
else:
    print("Mean and median are considerably different.")
```

```python
data = {
    "Product": ["Rice", "Oil", "Soap", "Shampoo", "Laptop",
                "Mouse", "Chair", "Notebook", "Pen", "Headphones"],
    "Category": ["Grocery", "Grocery", "Personal Care", "Personal Care",
                 "Electronics", "Electronics", "Furniture", "Stationery",
                 "Stationery", "Electronics"],
    "Quantity": [5, 2, 10, 3, 1, 4, 2, 12, 20, 2],
    "Unit_Price": [60, 150, 35, 120, 55000, 500, 3500, 50, 10, 1500],
    "Total_Sales": [300, 300, 350, 360, 55000, 2000, 7000, 600, 200, 3000]
}

df = pd.DataFrame(data)

# print(df)
Q1 = df["Total_Sales"].quantile(0.25)
Q3 = df["Total_Sales"].quantile(0.75)

print("Q1 :",Q1)
print("Q3 :", Q3)
IQR = Q3 - Q1
print("IQR :",IQR)

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[(df["Total_Sales"] < lower) | (df["Total_Sales"] > upper)]
print("Transactions with Outliers:")
print(outliers)
```

```python
data = {
    "Customer_ID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "Age": [25, 30, None, 45, 28, 35, 40, None, 32, 50],
    "Income": [30000, None, 45000, None, 38000, 52000, None, 41000, 47000, 60000],
    "Balance": [5000, 7000, 12000, 8000, None, 15000, 9000, 6000, 11000, 20000],
    "Credit_Score": [720, 680, 750, 700, None, 790, 650, 710, None, 760]
}

df = pd.DataFrame(data)

# print(df)
# print(df.isnull().sum())
# print(df.isnull().sum().idxmax())

# d. Replace missing Income values with median
df["Income"] = df["Income"].fillna(df["Income"].median())
df["Credit_Score"] = df["Credit_Score"].fillna(df["Credit_Score"].mean())
print("\nMissing Values After Filling:")
print(df.isnull().sum())
print("\nUpdated Income and Credit_Score:")
print(df[["Income", "Credit_Score"]])
```

```python
import pandas as pd

data = {
    "Date": [
        "2022-01-15", "2022-04-10", "2022-08-20", "2022-11-05",
        "2023-02-12", "2023-05-18", "2023-07-25", "2023-12-10",
        "2024-01-20", "2024-03-15", "2024-06-22", "2024-10-30"
    ],
    "Product": [
        "Rice", "Oil", "Soap", "Milk",
        "Rice", "Oil", "Shampoo", "Milk",
        "Soap", "Rice", "Oil", "Shampoo"
    ],
    "Quantity": [10, 5, 20, 15, 12, 8, 10, 18, 25, 14, 9, 12],
    "Sales": [5000, 3000, 2000, 1500,
              6000, 4500, 3500, 2000,
              2800, 7000, 5000, 4200]
}

df = pd.DataFrame(data)

# print(df)
df["Date"] = pd.to_datetime(df["Date"])

# print(df["Date"].dtype)
df["Year"] = df["Date"].dt.year
df["Month"] = df["Date"].dt.month

# print(df)

total_sales = df.groupby("Year")["Sales"].sum()
print("Total Sales Per Year:")
print(total_sales)
```

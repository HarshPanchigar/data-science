```python
import pandas as pd
import numpy as np

# 1. Create a Series from a Python list
sales_list = [12000, 15000, 18000, 14000, 16000]
series1 = pd.Series(sales_list)

# 2. Create a Series from a dictionary
sales_dict = {
    "Amit": 12000,
    "Raj": 15000,
    "Priya": 18000,
    "Neha": 14000,
    "Karan": 16000
}
series2 = pd.Series(sales_dict)

# 3. Create a Series from a NumPy array
sales_array = np.array([12000, 15000, 18000, 14000, 16000])
series3 = pd.Series(sales_array)

# Display all three Series
print("Series from Python List:")
print(series1)
print("Indexes:", series1.index)

print("\nSeries from Dictionary:")
print(series2)
print("Indexes:", series2.index)

print("\nSeries from NumPy Array:")
print(series3)
print("Indexes:", series3.index)
```

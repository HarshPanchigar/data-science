```python

import pandas as pd

# Create a DataFrame containing 8 students
data = {
    "Roll_No": [101, 102, 103, 104, 105, 106, 107, 108],
    "Name": ["Amit", "Raj", "Priya", "Neha", "Karan", "Riya", "Jay", "Kavya"],
    "Course": ["BCA", "IMSc IT", "BCA", "IMSc IT", "BCA", "IMSc IT", "BCA", "IMSc IT"],
    "Marks": [78, 65, 91, 56, 84, 73, 69, 88],
    "Attendance": [90, 85, 95, 80, 88, 92, 78, 96]
}

df = pd.DataFrame(data)

# Display complete DataFrame
print("Complete DataFrame:")
print(df)

# 1. Display the first 3 rows using loc
print("\nFirst 3 rows:")
print(df.loc[0:2])

# 2. Display the row having index 4
print("\nRow having index 4:")
print(df.loc[4])

# 3. Display Name and Marks for rows with index 2 to 5
print("\nName and Marks for index 2 to 5:")
print(df.loc[2:5, ["Name", "Marks"]])

# 4. Display rows at positions 1, 3, and 5 using iloc
print("\nRows at positions 1, 3, and 5:")
print(df.iloc[[1, 3, 5]])

# 5. Display the Name column using direct indexing
print("\nName column:")
print(df["Name"])
```

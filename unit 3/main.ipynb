```python

import pandas as pd

# Create 10 employee records
data = {
    "Emp_ID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "Name": ["Amit", "Raj", "Priya", "Neha", "Karan",
             "Riya", "Jay", "Kavya", "Dev", "Nisha"],
    "Department": ["IT", "HR", "Finance", "IT", "Sales",
                   "HR", "IT", "Sales", "Finance", "IT"],
    "Salary": [40000, 35000, 45000, 50000, 30000,
               38000, 55000, 42000, 48000, 60000],
    "Experience": [2, 3, 5, 4, 1, 3, 6, 4, 5, 7],
    "City": ["Ahmedabad", "Surat", "Mumbai", "Pune", "Delhi",
             "Ahmedabad", "Vadodara", "Surat", "Mumbai", "Pune"]
}

# Create DataFrame
df = pd.DataFrame(data)

# Save data to an Excel file
df.to_excel("employee_data.xlsx", index=False)

# Import the Excel file
df = pd.read_excel("employee_data.xlsx")

# Display first 5 records
print("First 5 Records:")
print(df.head())

# Display last 3 records
print("\nLast 3 Records:")
print(df.tail(3))

# Display shape
print("\nShape:")
print(df.shape)

# Display data types
print("\nData Types:")
print(df.dtypes)

# Display statistical summary
print("\nStatistical Summary:")
print(df.describe())
```

```python

import pandas as pd

# Create a dictionary of lists
data = {
    "Emp_ID": [101, 102, 103, 104, 105, 106, 107, 108],
    "Name": ["Amit", "Raj", "Priya", "Neha", "Karan", "Riya", "Jay", "Kavya"],
    "Department": ["IT", "HR", "Sales", "IT", "Finance", "HR", "IT", "Sales"],
    "Salary": [35000, 30000, 40000, 45000, 38000, 32000, 50000, 42000],
    "Experience": [2, 1, 3, 4, 3, 2, 5, 4]
}

# Create DataFrame
df = pd.DataFrame(data)

# 1. Complete DataFrame
print("Complete DataFrame:")
print(df)

# 2. Basic information
print("\nBasic Information:")
df.info()

# 3. Statistical summary
print("\nStatistical Summary:")
print(df.describe())

# 4. First 5 records
print("\nFirst 5 Records:")
print(df.head())

# 5. Last 3 records
print("\nLast 3 Records:")
print(df.tail(3))

# 6. Shape of DataFrame
print("\nShape:", df.shape)

# 7. Data types of all columns
print("\nData Types:")
print(df.dtypes)
```

```python
import pandas as pd

# Create a Pandas Series
marks = [78, 65, 91, 56, 84, 73, 69, 88]
series = pd.Series(marks)

# Create a DataFrame
data = {
    "Name": ["Amit", "Raj", "Priya", "Neha", "Karan", "Riya", "Jay", "Kavya"],
    "Subject": ["Python", "Python", "Python", "Python",
                "Python", "Python", "Python", "Python"],
    "Marks": marks
}
df = pd.DataFrame(data)

# Display both objects
print("Pandas Series:")
print(series)

print("\nDataFrame:")
print(df)

# Verify type, dimensions, and shape
print("\nSeries Type:", type(series))
print("Series Dimensions:", series.ndim)
print("Series Shape:", series.shape)

print("\nDataFrame Type:", type(df))
print("DataFrame Dimensions:", df.ndim)
print("DataFrame Shape:", df.shape)
```

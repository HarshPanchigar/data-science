```python
import numpy as np
import pandas as pd

df = pd.read_csv("ecommerce.csv")
print(df.head(5).to_string(index=False))

# Display shape, dtypes, head(), and tail().
print(df.shape)
print(df.dtypes)
print(df.head(),df.tail())
```

```python
# Rename Customer_Name to Customer and Payment_Mode to Payment.
df.rename(columns={"Customer_Name" : "Customer",
                   "Payment_Mode" : "Payment"})  
```

```python
df["Quantity"] = df["Quantity"].astype(float)
print(df)
```

```python
df['Total_Amount'] = df['Quantity'] * df['Price']
print(df)
```

```python
df[(df['Total_Amount'] > 2000) & (df['Quantity'] > 1)]
```

```python
df[(df["Region"] == "North") | (df["Region"] == "South")]
```
